```python
conn = sqlite3.connect("loans.db")
pd.read_sql_query("SELECT COUNT(*) AS n FROM loans;", conn)
```

```python
pd.read_sql_query("SELECT SUBSTR(issue_d, -4) AS year, COUNT(*) AS loans FROM loans GROUP BY year ORDER BY year;", conn)
```

```python
pd.read_sql_query("SELECT loan_status, COUNT(*) AS loans FROM loans GROUP BY loan_status ORDER BY loans DESC;", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * SUM(CASE WHEN loan_status = 'Charged Off' THEN 1 ELSE 0 END) / COUNT(*), 1) AS charge_off_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(AVG(int_rate), 2) AS avg_interest_pct,
       ROUND(100.0 * SUM(CASE WHEN loan_status = 'Charged Off' THEN 1 ELSE 0 END) / COUNT(*), 1) AS charge_off_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * (SUM(total_pymnt) - SUM(funded_amnt)) / SUM(funded_amnt), 1) AS net_return_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```
