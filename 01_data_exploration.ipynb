```python
conn = sqlite3.connect("loans.db")
pd.read_sql_query("SELECT COUNT(*) AS n FROM loans;", conn)
```

```python
pd.read_sql_query("SELECT SUBSTR(issue_d, -4) AS year, COUNT(*) AS loans FROM loans GROUP BY year ORDER BY year;", conn)
```

```python
pd.read_sql_query("SELECT loan_status, COUNT(*) AS loans FROM loans GROUP BY loan_status ORDER BY loans DESC;", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * SUM(CASE WHEN loan_status = 'Charged Off' THEN 1 ELSE 0 END) / COUNT(*), 1) AS charge_off_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(AVG(int_rate), 2) AS avg_interest_pct,
       ROUND(100.0 * SUM(CASE WHEN loan_status = 'Charged Off' THEN 1 ELSE 0 END) / COUNT(*), 1) AS charge_off_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * (SUM(total_pymnt) - SUM(funded_amnt)) / SUM(funded_amnt), 1) AS net_return_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * (SUM(total_pymnt) - SUM(funded_amnt)) / SUM(funded_amnt), 1) AS net_return_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
  AND SUBSTR(issue_d, -4) <= '2014'
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
pd.read_sql_query("""
SELECT grade,
       COUNT(*) AS loans,
       ROUND(100.0 * (SUM(total_pymnt) - SUM(funded_amnt)) / SUM(funded_amnt), 1) AS net_return_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
  AND TRIM(term) = '36 months'
  AND SUBSTR(issue_d, -4) <= '2014'
GROUP BY grade
ORDER BY grade;
""", conn)
```

```python
import matplotlib.pyplot as plt

grade_df = pd.read_sql_query("""
SELECT grade,
       ROUND(100.0 * SUM(CASE WHEN loan_status = 'Charged Off' THEN 1 ELSE 0 END) / COUNT(*), 1) AS charge_off_pct,
       ROUND(100.0 * (SUM(total_pymnt) - SUM(funded_amnt)) / SUM(funded_amnt), 1) AS net_return_pct
FROM loans
WHERE loan_status IN ('Fully Paid', 'Charged Off')
  AND TRIM(term) = '36 months'
  AND SUBSTR(issue_d, -4) <= '2014'
GROUP BY grade
ORDER BY grade;
""", conn)

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(grade_df["grade"], grade_df["charge_off_pct"], label="Charge-off rate (%)")
ax.plot(grade_df["grade"], grade_df["net_return_pct"], marker="o", color="black", label="Net return (%)")
ax.set_xlabel("Loan grade")
ax.set_ylabel("Percent")
ax.set_title("Charge-off rate vs. net return by grade (36-month loans, 2007-2014)")
ax.legend()
plt.show()
```
